```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\Monika\Downloads\titanic_raw.csv")

print("Dataset loaded successfully!")
df.head()
```

```python
print("Shape of dataset:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("Number of duplicate records:", duplicates)
```

```python
df = df.drop_duplicates()

print("Duplicates removed.")
print("New shape:", df.shape)
```

```python
print("Unique values in Sex:")
print(df["Sex"].unique())

print("\nUnique values in Embarked:")
print(df["Embarked"].unique())
```

```python
df["Sex"] = df["Sex"].astype(str).str.strip().str.lower()

df["Sex"] = df["Sex"].replace({
    "male": "Male",
    "female": "Female"
})

df["Embarked"] = df["Embarked"].astype(str).str.strip().str.upper()
```

```python
df["Age"] = df.groupby("Pclass")["Age"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
print("Missing Age values:", df["Age"].isnull().sum())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
df["PassengerId"] = df["PassengerId"].astype(int)
df["Survived"] = df["Survived"].astype(int)
df["Pclass"] = df["Pclass"].astype(int)

df["Age"] = df["Age"].astype(float)
df["SibSp"] = df["SibSp"].astype(int)
df["Parch"] = df["Parch"].astype(int)
df["Fare"] = df["Fare"].astype(float)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate records after cleaning:")
print(df.duplicated().sum())

print("\nData types after cleaning:")
print(df.dtypes)
```
